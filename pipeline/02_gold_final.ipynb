```python
import pandas as pd   #испортируем библиотеку
```

```python
gold = pd.read_excel("goldenset(3).xlsx")      #загружаем данные: прошлый голденсет и таблицу с релевантными чанками
markup = pd.read_excel("gold_markup8.xlsx")

print("gold:", gold.shape)
print("markup:", markup.shape)
```

```python
bad_ids = [40, 41, 42, 43, 44]         #удаляем проблемные вопросы, т.к они не попали в чанки, к примеру, из-за отстуствия данных

gold = gold.drop(index=bad_ids)
markup = markup[~markup["question_id"].isin(bad_ids)]

print("gold после:", gold.shape)          #смотрим что получилось
print("markup после:", markup.shape)
```

```python
relevant = markup[markup["is_relevant"] == 1][["question_id", "chunk_id"]]   #отсавляем только релевантные чанки, выбранные в процессе ручной разметки

print("relevant:", relevant.shape)
```

```python
gold_final = gold.merge(       #присоединяем chunk_id к исходному голдесету по question_id, используем индексы голденсета как ключ
    relevant,
    left_index=True,      # используем индекс gold
    right_on="question_id",
    how="left"
)
```

```python
missing = gold_final["chunk_id"].isna().sum()   #проверяем, что у каждого вопроса появился chunk_id
print("пропусков:", missing)         #должно быть 0
```

```python
gold_final.to_csv("gold_final.csv", index=False)

print("gold_final.csv готов")
```
